```python
df=spark.table("capstone4_retail.retail_orders.bronze_sale_events")
df.printSchema()
print(df.count())
```

```python
from pyspark.sql.functions import col
print(df.filter(col("sales_channel").isNull()).count())
print(df.filter(col("promotion_id").isNull()).count())
```

```python
distinct=df.select("sale_id").distinct().count()
total=df.count()
print(total-distinct)

```

```python
df.select("sales_amount","quantity").describe().show()
```

```python
metric_df = df.withColumn(
    "sales_amount_num",
    col("sales_amount").cast("decimal(18,2)")
)
```

```python
metric_df.printSchema()
```

```python
metric_df = spark.table("capstone4_retail.retail_orders.bronze_sale_events").withColumn(
    "sales_amount_num",
    col("sales_amount").cast("decimal(18,2)")
)
print(metric_df.filter(col("sales_amount_num")>1000).count())
```

```python
metric_df = spark.table("capstone4_retail.retail_orders.bronze_sale_events").withColumn(
    "quantity_num",
    col("quantity").cast("integer")
)
print(metric_df.filter(col("quantity_num")>1000).count())
```
