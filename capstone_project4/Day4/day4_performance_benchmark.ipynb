```python
# DAY 4 - PERFORMANCE BENCHMARK

task_performance = [
    ("bronze_ingestion", 47.1),
    ("jdbc_ingestion", 49.0),
    ("sales_silver", 80.0),
    ("inventory_silver", 26.0),
    ("returns_silver", 39.5),
    ("master_silver", 92.0),
    ("gold_dimensions", 24.2),
    ("gold_facts", 28.5),
    ("gold_business", 29.7),
    ("gold_validation", 23.2)
]

performance_df = spark.createDataFrame(
    task_performance,
    ["task_name", "duration_seconds"]
)

display(
    performance_df.orderBy("duration_seconds", ascending=False)
)
```

### Identify the slowest tasks



```python
# Identify slowest tasks

display(
    performance_df
    .orderBy("duration_seconds", ascending=False)
    .limit(3)
)
```

### Calculate total task execution time

```python
from pyspark.sql.functions import sum, avg, max

performance_df.select(
    sum("duration_seconds").alias("total_task_seconds"),
    avg("duration_seconds").alias("average_task_seconds"),
    max("duration_seconds").alias("maximum_task_seconds")
).show()
```
