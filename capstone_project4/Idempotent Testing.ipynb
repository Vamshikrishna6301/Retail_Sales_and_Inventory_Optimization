```python
%sql
SELECT 'Bronze Sales' AS table_name, COUNT(*) AS total
FROM capretailnew.bronze.bronze_sales_order_lines

UNION ALL

SELECT 'Bronze Inventory', COUNT(*)
FROM capretailnew.bronze.bronze_inventory_movements

UNION ALL

SELECT 'Bronze Returns', COUNT(*)
FROM capretailnew.bronze.bronze_sales_returns

UNION ALL

SELECT 'Gold Sales Fact', COUNT(*)
FROM capretailnew.gold.gold_fact_sales

UNION ALL

SELECT 'Gold Returns Fact', COUNT(*)
FROM capretailnew.gold.gold_fact_returns

UNION ALL

SELECT 'Gold Inventory Fact', COUNT(*)
FROM capretailnew.gold.gold_fact_inventory_movements;
```
