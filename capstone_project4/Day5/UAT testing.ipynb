##What is the total net sales for each product category? Show the categories in descending order of net sales.

```python
%sql
SELECT
    category,
    ROUND(SUM(sales_after_discount_before_tax), 2) AS total_net_sales
FROM capretailnew.gold.gold_daily_sales_summary
GROUP BY category
ORDER BY total_net_sales DESC;
```

##Which are the top 5 stores based on total net sales? Show the store name and total net sales in descending order.

```python
%sql
SELECT
    store_id,
    store_name,
    ROUND(SUM(sales_after_discount_before_tax), 2) AS total_net_sales
FROM capretailnew.gold.gold_daily_sales_summary
GROUP BY store_id, store_name
ORDER BY total_net_sales DESC
LIMIT 5;
```

##Show the monthly net sales trend across all stores and product categories. Display each month in chronological order and include total net sales.

```python
%sql
SELECT
    DATE_FORMAT(sales_date, 'yyyy-MM') AS sales_month,
    ROUND(SUM(sales_after_discount_before_tax), 2) AS total_net_sales,
    COUNT(DISTINCT sales_date) AS days_with_sales
FROM capretailnew.gold.gold_daily_sales_summary
GROUP BY DATE_FORMAT(sales_date, 'yyyy-MM')
ORDER BY sales_month;
```
